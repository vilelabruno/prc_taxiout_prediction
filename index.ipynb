```python
from pyspark.sql import SparkSession

# Initialize Spark session
spark = SparkSession.builder \
    .appName("Check Cluster") \
    .getOrCreate()

# Access SparkContext
sc = spark.sparkContext

# Get the list of all executor nodes and their memory status
executor_memory_status = sc._jsc.sc().getExecutorMemoryStatus()

# Iterate through the Java Map
for executor in executor_memory_status.keySet():
    status = executor_memory_status.get(executor)
    print(f"Executor: {executor}, Status: {status}")
```
